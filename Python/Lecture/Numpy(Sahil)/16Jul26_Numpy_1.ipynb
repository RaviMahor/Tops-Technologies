```python
"""
1. Create an array from 1 to 15.
Split it into 5 equal parts.
""" #AI
import numpy as np 

# 1 se 15 tak ka array banana
arry1 = np.arange(1, 16)

# 5 equal parts mein split karna
parts = np.split(arry1, 5)

print("Original Array:", arry1)
print("Split Parts:")
for part in parts:
    print(part)
```
