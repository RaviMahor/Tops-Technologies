```python
# Operation in Numpy Array
a = np.array([10,20,30])
b = np.array([10,20,30])

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a**b)
print(a%b)
```

```python
c = np.array([1,2,3,4])
d = np.array([2,2,2,2])

print(c==d)
print(c!=d)
print(c<d)
print(c>d)
print(c>=d)
print(c>=d)
```

```python
e = np.array([True, False, True])
f = np.array([True, True, False])

print(np.logical_and(e,f))
print(np.logical_or(e,f))
print(np.logical_not(e))
```

```python
# Mathematical Operations
a = np.array([4,9,16])
b = np.array([4,9,1])

print(np.add(a,b))
print(np.subtract(a,b))
print(np.multiply(a,b))
print(np.divide(a,b))
print(np.power(a,b))
print(np.mod(a,b))
print(np.sqrt(a))
print(np.exp(a))
print(np.log(a))
print(np.log10(a))
```
