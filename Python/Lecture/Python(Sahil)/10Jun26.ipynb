```python
""" 
9. Write a function called Calculate_discount that takes price and rate as arguments 
and returns the final price after discount.
"""
def calculate_discount(price, rate):
    discount_amount = price * (rate / 100)
    final_price = price - discount_amount
    return final_price
price = float(input("Enter the original price: "))
rate = float(input("Enter the discount rate (in percentage): "))
final_price = calculate_discount(price, rate) # calling the function and storing the result in a variable
print("The final price after discount is: ", final_price)
```

```python
""" 
13. Write a function called count_digits that takes a number as argument and
returns the number of digits in that number.
"""
def count_digits(number):
    # 1. Negative sign ko ignore karne ke liye abs() use kiya
    # 2. str() se text me convert kiya aur len() se characters count kiye
    return len(str(abs(number)))

# Test standard cases:
print(count_digits(12345))  # Output: 5
print(count_digits(0))      # Output: 1
print(count_digits(-987))   # Output: 3
```

```python
""" 
13. Write a function called count_digits that takes a number as argument and
returns the number of digits in that number.
"""
def count_digits(number):
    number = abs(number) # Negative value ko positive karne ke liye
    
    if number == 0:
        return 1
        
    count = 0
    while number > 0:
        count += 1
        number = number // 10 # Ek digit ko right side se hatane ke liye (integer division)
    return count

# Test cases:
print(count_digits(5020))  # Output: 4
```

```python
""" 
14. Write a function called total_sales that takes a list of 
sales values as argument and returns the total sales.
"""
def total_sales(sales):
    total = 0
    for i in sales:
        total = i + total
    return total
    
sales = [10,20,30,40]
total_sales(sales)
```

```python
""" 
15.  Write a function called clean_product_name that takes
a sring as argument and performs the following operations
convert the string to lowercase
remove extra spaces from beginning and end
replace any dash with a space 
return the cleaned product name
"""
def clean_product_name(product_name):
    # 1. Sabhi characters ko lowercase me badlein
    name_lower = product_name.lower()
    
    # 2. Shuruat aur aakhiri ke extra spaces hatayein
    name_stripped = name_lower.strip()
    
    # 3. Dash '-' ko space ' ' se replace karein
    final_name = name_stripped.replace('-', ' ')
    
    return final_name

# Test cases:
raw_name = "   Smart-Phone-123   "
print("Before:", raw_name)
print("After :", clean_product_name(raw_name))  
# Output: "smart phone 123"
```
