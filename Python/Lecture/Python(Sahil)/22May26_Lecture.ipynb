```python
# Creating tuple
coordinates = (10.0, 20.0)
print(coordinates)
print(type(coordinates))

# indexing
print("First element:", coordinates[0])  # Output: 10.0

# Slicing
print("Sliced tuple:", coordinates[0:2])  # Output: (10.0, 20.0)
```

```python
# Mulablity Demo
point = (5.0, 10.0)

# This will give error if uncommented because tuples are immutable
point[0] = 15.0  # TypeError: 'tuple' object does not support item assignment
print("Tuple after modification attempt:", point)
```

```python
# tuple methods
my_tuple = (10, 20, 30, 40, 50, 20, 30)

# count method
# Returns number of times value appears in the tuple
print("Count of 20 in tuple:", my_tuple.index(20))  # Output: 2
```

```python
A = {1, 2, 3}  
B = {3, 4, 5}

# Add 
A.add(6)  # This will give error because tuples do not have an add method
print("After adding 6 to A:", A)

#remove Element 
A.remove(2)  # This will give error because tuples do not have a remove method
print("After removing 2 from A:", A)

# discard Element (no error if element not found
A.discard(100)  # This will give error because tuples do not have a discard method
print("After discarding 100 from A:", A)

# pop() - removes and returns an arbitrary element
A.pop()  # This will give error because tuples do not have a pop method
print("After popping an element from A:", A)

# union
print("Union of A and B:", A.union(B))  # This will give error because tuples do not have a union method
# intersection
print("Intersection of A and B:", A.intersection(B))  # This will give error because tuples do not have an intersection method
# difference
print("Difference of A - B:", B.difference(A))  # This will give error  
```
