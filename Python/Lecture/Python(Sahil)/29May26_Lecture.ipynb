```python
ppl_age = int(input("Enter your age: "))
if ppl_age >= 13:
    if ppl_age < 18:
        print("You are a minor.")
    elif ppl_age < 65:
        print("You are an adult.")
    else:
        print("You are a senior.")
else:
    print("You are a child.")        
```

```python
revenue = float(input("Enter your revenue: "))
target_revenue = 100000.0
region = input("Enter your region (North, South, East, West): ")
if revenue >= target_revenue:
    if region == "North":
        print("Congratulations! You have met the target revenue for the North region.")
    elif region == "South":
        print("Congratulations! You have met the target revenue for the South region.")
    elif region == "East":
        print("Congratulations! You have met the target revenue for the East region.")
    elif region == "West":
        print("Congratulations! You have met the target revenue for the West region.")
    else:
        print("Invalid region entered.")
else:
    print("You have not met the target revenue.")
```
