```python
"""
4. Take a number as input. 
Print its multiplication table (up to 10) using while loop.
"""
num = int(input("Enter a number: "))
i = 1
while i <= 10: # while i <= 10: is ka matlab hai ke jab tak i 10 se chota ya barabar hai, loop chalta rahega
    print(num, "x", i, "=", num * i)
    i = i + 1
```

```python
"""
4. Take a number as input. 
Print its multiplication table (up to 10) using for loop.
"""
num = int(input("Enter a number: "))
for i in range(1, 11): # for i in range(1, 11): is ka matlab hai ke i 1 se start hoga aur 10 tak chalega, 11 include nahi hoga
    print(num, "x", i, "=", num * i)
    
```

```python
"""
5. Calculate the sum of numbers from 1 to 50 using a while loop. 
"""
total = 0
num = 1
while num <= 50: # while num <= 50: is ka matlab hai ke jab tak num 50 se chota ya barabar hai, loop chalta rahega
    total = total + num # total = total + num is ka matlab hai ke total mein num ko add karna hai
    num = num + 1 # num = num + 1 is ka matlab hai ke num ko 1 se barhana hai
print("total:", total)
```

```python
""" 
6. Take numbers from user continuously.
Stop when user enters 0.
Print the total sum of all entered numbers. 
"""
total = 0
while True: # while True: is ka matlab hai ke loop hamesha chalta rahega, jab tak break statement nahi aata
    num = int(input("Enter a number (0 to stop): "))
    if num == 0: # if num == 0: is ka matlab hai ke agar user ne 0 enter kiya hai, to loop ko stop karna hai
        break
    total = total + num # total = total + num is ka matlab hai ke total mein num ko add karna hai
print("Total sum:", total)
```

```python
""" 
7. Count how many digits are there in a given number using while loop.
"""
number = int(input("Enter a number: ")) 
count = 0 
while number > 0: # while number > 0: is ka matlab hai ke jab tak number 0 se bada hai, loop chalta rahega
    number = number // 10 
    count = count + 1
print("Number of digits:", count)
```

```python
""" 
8. Reverse a number using while loop. 
Example: Input: 1234, Output: 4321
"""
number = int(input("Enter a number: "))
reverse = 0
while number > 0: # while number > 0: is ka matlab hai ke jab tak number 0 se bada hai, loop chalta rahega
    digit = number % 10 # digit = number % 10 is ka matlab hai ke number ko 10 se divide karna hai aur uska remainder lena hai, jisse last digit mil jayega
    reverse = reverse * 10 + digit # reverse = reverse * 10 + digit is ka matlab hai ke reverse ko 10 se multiply karna hai aur usme digit ko add karna hai, jisse reverse number ban jayega
    number = number // 10 # number = number // 10 is ka matlab hai ke number ko 10 se divide karna hai aur uska integer part lena hai, jisse last digit remove ho jayega
print("Reversed number:", reverse)
```

```python
""" 
9. keep asking the user to enter a password.
Stop only when the correct password is entered.
"""
correct_password = "123"
while True: # while True: is ka matlab hai ke loop hamesha chalta rahega, jab tak break statement nahi aata
    password = input("Enter the password: ")
    if password == correct_password: # if password == correct_password: is ka matlab hai ke agar user ne sahi password enter kiya hai, to loop ko stop karna hai
        print("Correct password entered. Access granted.")
        break
    else:
        print("Incorrect password. Try again.") 
```

```python
""" 
10. Find the factorial of a number using while loop.
Example: 5! = 5 x 4 x 3 x 2 x 1 = 120
"""
number = int(input("Enter a number: "))
factorial = 1
while number > 0: # while number > 0: is ka matlab hai ke jab tak number 0 se bada hai, loop chalta rahega
    factorial = factorial * number # factorial = factorial * number is ka matlab hai ke factorial ko number se multiply karna hai, jisse factorial ban jayega
    # print(factorial)
    number = number - 1 # number = number - 1 is ka matlab hai ke number ko 1 se ghataana hai, jisse next iteration mein next number aayega
    # print(number)
print("Factorial:", factorial)
```
