```python
""" 
15. Create two sets: A = {1,2,3,4} and B = {3,4,5,6}. Find:
- Union
- Intersection
- Difference A - B
"""
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
union = A.union(B)  # Union of A and B
intersection = A.intersection(B)  # Intersection of A and B
difference = A.difference(B)  # Difference of A - B
print(f"Union: {union}")  # Output: {1, 2, 3, 4, 5, 6}
print(f"Intersection: {intersection}")  # Output: {3, 4}
print(f"Difference A - B: {difference}")  # Output: {1, 2}
```

```python
""" 
18. Given a list of numbers: [10, 20, 30, 40, 50]
- Convert it to tuple
- Convert tuple to set
- Convert set back to list 
Print each step.
"""
numbers = [10, 20, 30, 40, 50]
# Convert list to tuple
numbers_tuple = tuple(numbers)
print(f"Tuple: {numbers_tuple}")  # Output: (10, 20, 30, 40, 50)
# Convert tuple to set
numbers_set = set(numbers_tuple)
print(f"Set: {numbers_set}")  # Output: {10, 20, 30, 40, 50}
# Convert set back to list
numbers_list = list(numbers_set)
print(f"List: {numbers_list}")  # Output: [10, 20, 30, 40, 50]  
```

```python
""" 
19. Given: products = { "Pen": 10, "Notebook": 50, "Pencil": 5 }
Calculate total price of all products.
If total > 60 print "High Inventory Value" 
Else print "Normal Inventory Value"
"""
products = {"Pen": 10, "Notebook": 50, "Pencil": 5}
total_price = sum(products.values())  # Calculate total price of all products
print(f"Total Price: {total_price}")  # Output: Total Price: 65
if total_price > 60:
    print("High Inventory Value")
else:
    print("Normal Inventory Value")
```

```python
""" 
22. Given a tuple: (10, 20, 30, 20,40, 20)
- Count how many times 20 appears.
- Find index of 30.
"""
my_tuple = (10, 20, 30, 20, 40, 20)
count_20 = my_tuple.count(20)  # Count how many times 20 appears
index_30 = my_tuple.index(30)  # Find index of 30
print(f"20 appears {count_20} times")  # Output: 20 appears 3 times
print(f"Index of 30: {index_30}")  # Output: Index of 30: 2
```
