```python


# Q2
# Take input for marks in 3 subjects as string,
# convert them into float,
# and print average marks.
marks1 = input("Enter marks for subject 1: ")
marks2 = input("Enter marks for subject 2: ")
marks3 = input("Enter marks for subject 3: ")
marks1 = float(marks1)
marks2 = float(marks2)
marks3 = float(marks3)
Total_marks = marks1 + marks2 + marks3
print(f"Total marks: {Total_marks}")
average_marks = (marks1 + marks2 + marks3) / 3
print(f"Average marks: {average_marks}")

```

```python
#>>>>>>
# Q3
# Create a variable:
# product = "   HP-Laptop   "
# Perform:
# - remove spaces
# - convert to lowercase
# - replace "-" with space
# Print final output.
product = "   HP-Laptop   "
product = product.strip()  # Remove spaces
print(product)
product = product.lower()  # Convert to lowercase
print(product)
product = product.replace("-", " ")  # Replace "-" with space
print(product)
```

```python

# Q4
# Take user input for first_name and last_name.
# Print full name in uppercase.
first_name = input("Enter your first name: ")
last_name = input("Enter your last name: ")
full_name = f"{first_name} {last_name}"
print(full_name.upper())

```

```python

# Q5
# Create a string:
# course = "ArtificialIntelligence"
# Print:
# a) First 10 characters
# b) Last 5 characters
# c) Every second character using slicing
course = "ArtificialIntelligence"
print(course[:10])  # First 10 characters
print(course[-5:])  # Last 5 characters
print(course[::2])  # Every second character using slicing
```

```python


# Q7
# Create a variable:
# price = "2500"
# Convert it into float,
# add 18% GST,
# and print final amount.
price = "2500"
price = float(price)  # Convert to float
gst = price * 0.18  # Calculate GST
final_amount = price + gst  # Add GST to original price
print(f"Final amount after adding 18% GST: {final_amount}")

```
