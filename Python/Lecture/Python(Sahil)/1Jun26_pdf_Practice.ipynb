```python
""" 
1. Write a for loop that prints numbers from 1 to 5.
"""
for i in range(1, 6): # Start at 1, go up to 5 (exclusive), so it will print 1, 2, 3, 4, 5
    print(i)
```

```python
""" 
2. Write a for loop that prints all even numbers from 1 to 20.
"""
for i in range(2, 21, 2): # Start at 2, go up to 20, and step by 2 to get even numbers
    print(i)
```

```python
""" 
3. Write a for loop that prints each element in the list: [10, 20, 30, 40, 50].
"""
numbers = [10, 20, 30, 40, 50]
for num in numbers: # Loop through each element in the list and print it
    print(num)
```

```python
""" 
4. Write a for loop that finds the sum of numbers from 1 to 10.
"""
total = 0

for i in range(1, 11): # Loop through numbers from 1 to 10 (inclusive)
    total += i # This is the same as total = total + i
print(total)
```

```python
""" 
5. Write a for loop that prints the multiplication table of 7.
"""
for i in range(1, 11): # Loop through numbers from 1 to 10 (inclusive)
    print(f"7 x {i} = {7 * i}") # This uses an f-string to format the output, showing the multiplication table of 7
```

```python
""" 
6. Write a for loop that counts how many numbers in the list [2,5,8,11,14] are greater than 6.
"""
numbers = [2,5,8,11,14]
count = 0
for n in numbers:
    if n > 6:
        count += 1
print(count)
```

```python
""" 
7. Write a for loop that prints each letter in the word "Python".
"""
word = "Python"
for letter in word: # Loop through each character in the string "Python"                
    print(letter)       # This will print each letter on a new line
```

```python
""" 
8. Write a for loop that prints numbers from 10 down to 1.
"""
for i in range(10, 0, -1): # Start at 10, go down to 1 (exclusive), and step by -1 to count backwards
    print(i)
```

```python
""" 
9. Write a for loop that finds the largest number in the list [3, 9, 1, 6, 4].
"""
numbers = [3, 9, 1, 6, 4]
largest = numbers[0] # Initialize largest with the first number in the list
for n in numbers: # Loop through each number in the list
    if n > largest: # If the current number is greater than the largest found so far
        largest = n # Update largest to be the current number
print(largest)
```

```python
""" 
10. Write a for loop that prints the square of numbers from 1 to 10.
"""
for i in range(1, 11): # Loop through numbers from 1 to 10 (inclusive)
    print(i**2)        # Print the square of the current number
```
