```python
""" 
1. Find the sum of digits of a given positive integer.
   Example: 1234 → 10 
"""
def sum_of_digits(n):
    # Handle negative numbers (optional but good practice)
    n = abs(n)
    
    # Base case: if the number is a single digit
    if n < 10:
        return n
    
    # Recursive case: last digit + sum of digits of the rest
    return (n % 10) + sum_of_digits(n // 10)

# Test the function
print(sum_of_digits(1234))
```

```python
""" 
2. Find the product of digits of a given positive integer.
   Example: 1234 → 24
"""
def product_of_digits(n):
    n = abs(n)
    
    # Base case: if the number is a single digit
    if n < 10:
        return n
    
    # Recursive case: last digit * product of digits of the rest
    return (n % 10) * product_of_digits(n // 10)

# Test the function
print(product_of_digits(1234))
```

```python
""" 
3. Count the number of zeros present in a given integer.
   Example: 102030 → 3 
"""
def count_zeros(n):
    n = abs(n)
    
    # 1. Base Case: Agar single digit number bacha hai (0 se 9 tak)
    if n < 10:                                                        # Base Case condition
        if n == 0:
            return 1  # Agar number 0 hai, toh hume 1 zero mila.
        else:
            return 0  # Agar number 1 se 9 ke beech hai, toh 0 zero mile.
            
    # 2. Last digit check karna:
    if n % 10 == 0:
        current_zero = 1  # Agar last digit 0 hai (jaise 10, 20), toh current_zero = 1.
    else:
        current_zero = 0  # Agar last digit 0 nahi hai, toh current_zero = 0.
        
    # 3. Baaki bache number ko check karna (Recursive Case):
    return current_zero + count_zeros(n // 10)

# Test the function
print(count_zeros(102030))
```

```python
""" 
4. Reverse a given number using recursion.
   Example: 1234 → 4321
"""
def reverse_number(n):
    sign = -1 if n < 0 else 1
    n = abs(n)
    
    # Base case: if single digit
    if n < 10:
        return n * sign
    
    # Calculate position of the last digit
    num_digits = len(str(n))
    last_digit = n % 10
    remaining = n // 10
    
    # Recursive case
    return sign * (last_digit * (10 ** (num_digits - 1)) + abs(reverse_number(remaining)))

# Test the function
print(reverse_number(1234))
```
