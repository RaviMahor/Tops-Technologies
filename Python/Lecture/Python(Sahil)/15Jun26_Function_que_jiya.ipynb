```python
""" 
6. Create a function named is_even(num) that returns True if the number is even, otherwise False. 
"""
def is_even(num):
    return num % 2 == 0

print(is_even(4))
print(is_even(7))
```

```python
""" 
7. Create a function named find_largest(a, b) that returns the larger of two numbers. 
"""
def find_largest(a, b):
    if a > b:
        return a
    else:
        return b

result = find_largest(15, 23)
print(result)
```

```python
""" 
9. Create a function named count_characters(text) that 
returns the total number of characters present in a string. 
"""
def count_characters(text):
    return len(text)

result = count_characters("Hello Jiya")
print(result)
```

```python
""" 
10. Create a function named calculate_factorial(n) that returns 
the factorial of a given number using a function.
"""
def calculate_factorial(n):
    # Base case
    if n <= 1:
        return 1
    
    # Recursive case
    return n * calculate_factorial(n - 1)

result = calculate_factorial(5)
print(result)
```
