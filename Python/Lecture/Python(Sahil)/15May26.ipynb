```python
# Every character in a string has an index, starting from 0 for the first character. You can access characters using their index.
City = "New York"

# N e w   Y o r k
# 0 1 2   3 4 5 6
# -6 -5 -4 -3 -2 -1

print(City[0])  # Output: 'N'
print(City[1])  # Output: 'e'       
print(City[-1])  # Output: 'k' (Last character)
```

```python
# Slicing = exteracting part of a string

test = "python programming"
print(test[0:6])  # Output: 'python' (from index 0 to 5)
print(test[:])  # Output: 'python programming' (from index "" to "")
print(test[7:])  # Output: 'programming' (from index 7 to the end)
print(test[:6])  # Output: 'python' (from index 0 to 5)

```

```python
brand = "Apple Inc."
print(brand.upper())  # Output: 'APPLE INC.'
```

```python
brand = "Apple Inc."
print(brand.lower())  # Output: 'apple inc.'
```

```python
# replace() method is used to replace a specified phrase with another specified phrase.

product = "iPhone 13 Pro"
print(product.replace("Pro", "Max"))  # Output: 'iPhone 13 Max'
```

```python
#split() method is used to split a string into a list where each word is a list item.

sentence = "Hello, how are you?"
words = sentence.split("o")  # Split by whitespace
print(words)
```

```python
#split() method is used to split a string into a list where each word is a list item.

sentence = "Hello, how are you?"
words = sentence.split(" ")  # Split by whitespace
print(words)
```

```python
name = "John Doe"
age = 30
print("My name is", name, "and I am", age, "years old.")  # Output: 'My name is John Doe and I am 30 years old.'
```

```python
# f-string (formatted string literals) allows you to embed expressions inside string literals, using curly braces {}.
name = "John Doe"
age = 30
print(f"My name is {name} and I am {age} years old.")  # Output: 'My name is John Doe and I am 30 years old.'
```
