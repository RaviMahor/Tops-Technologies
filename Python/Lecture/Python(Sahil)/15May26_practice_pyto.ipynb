```python

# 1. Take the complete address input from the user
print("Please enter your complete address separated by commas.")
print("Format: House No, Street, City, State, Country, Pin Code")
user_address = input("Enter address: ")

```

```python
# 2. Split (slice) the string into a list based on the commas
address_parts = user_address.split(",")

```

```python
# 3. Check if the user entered all 6 required parts
if len(address_parts) == 6:
    # .strip() removes any accidental extra spaces the user typed
    house_no = address_parts[0].strip()
    street_name = address_parts[1].strip()
    city = address_parts[2].strip()
    state = address_parts[3].strip()
    country = address_parts[4].strip()
    pin_code = address_parts[5].strip()

    # 4. Display the sliced components
    print("\n--- Sliced Address Components ---")
    print(f"1. House/Flat No.: {house_no}")
    print(f"2. Street Name   : {street_name}")
    print(f"3. City          : {city}")
    print(f"4. State         : {state}")
    print(f"5. Country       : {country}")
    print(f"6. Pin Code      : {pin_code}")
else:
    print("\nError: Please make sure to provide all 6 parts separated by commas.")
```
