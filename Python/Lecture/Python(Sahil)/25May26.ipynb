```python
product = {
    "Laptop": 50000,
    "Mobile": 20000,
}

print(product)
```

```python
# Access
print(product["Laptop"])
```

```python
# Add
product["Table"] = 15000
print(product)
```

```python
# Update
product["Mobile"] = 22000
print(product)
```

```python
# Delete
del product["Table"]
print(product)
```

```python
# Keys
print(product.keys())
```

```python
# Values
print(product.values())
```

```python
# Items
print(product.items())
```

```python
# revenue = 12,00,000
revenue = int(input("Enter Revenue:"))

# target = 13,00,000
target = int(input("Enter Target:"))

# Person Salary = 15,000
Person = int(input("Enter salary"))

if revenue > target:
    print("Achieved")
    Person = Person + 2000
    print(Person)
if revenue < target:
    print("Not achieved")
    Person = Person - 2000
    print(Person)
```
