```python
"""
5. Write a program that uses try, except, and else.
Take a number as input and print its square only if no error occurs.
""" #AI
try:
    # Take a number as input and convert to float
    user_input = input("Enter a number: ")
    
    number = float(user_input)
except ValueError:
    # Handle the error if input is not a valid number
    print("Error: Please enter a valid numerical value.")
else:
    # This runs only if no error occurred in the try block
    square = number ** 2
    
    # If the square is a whole number, format it as an integer
    if square.is_integer():
        square = int(square)
        
    print(f"The square of the number is: {square}")
```

```python
"""
6. Write a program that demonstrates the finally block.
Print a message saying "Program completed" 
whether an error occurs or not.
"""
try:
    # Ask the user for a number to divide 100
    user_input = input("Enter a number to divide 100: ")
    number = float(user_input)
    result = 100 / number
    print(f"Result: {result}")
except ValueError:
    print("Error: Please enter a valid number.")
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
finally:
    # This block will run no matter what (with or without errors)
    print("Program completed")
```

```python
"""
8. Write a program that asks the user for two numbers and 
safely performs division. Handle both invalid input and division by 
zero using multiple except blocks.
""" #AI
try:
    # Ask the user for two numbers
    num1 = float(input("Enter the first number (numerator): "))
    num2 = float(input("Enter the second number (denominator): "))
    
    # Perform the division
    result = num1 / num2
    
except ValueError:
    # Runs if the user inputs text instead of a number
    print("Error: Invalid input. Please enter valid numbers only.")
    
except ZeroDivisionError:
    # Runs if the user tries to divide by zero
    print("Error: Division by zero is not allowed.")
    
else:
    # Runs only if no errors occurred
    if result.is_integer(): # check .0 is present than its true
        result = int(result)
    print(f"The result of division is: {result}")
```
