```python
"""
1. Write a recursive fuction called countdown that takes a number n 
and prints numbers from n down to 1.
"""
def countdown(n):
    # 1. Base Case: Agar n 0 ya usse chota ho jaye to ruk jao
    if n <= 0:
        return
    
    # 2. Pehle current number print karo
    print(n)
    
    # 3. Recursive Call: Agla number (n - 1) print karne ke liye fir se call karo
    countdown(n - 1)

# Test the function:
countdown(5)
```

```python
""" 
2. Write a recursive function called print_numbers
that prints numbers from 1 to n.
"""
def print_numbers(n):
    # Base Case: Agar n 0 ya usse chota ho jaye to ruk jao
    if n <= 0:
        return
    
    # 1. Pehle (n - 1) tak ke numbers print karne ke liye recursive call karo
    print_numbers(n - 1)
    
    # 2. Jab andar wali call se wapas aayein, tab current number print karo
    print(n)

# Test the function:
print_numbers(5)
# Output:
# 1
# 2
# 3
# 4
# 5
```

```python
""" 
3. Write a recursive function called 
factorial that calculates the factorial of a number.
"""
def factorial(n):
    # 1. Base Case: Agar n 0 ya 1 hai, to factorial 1 hota hai
    if n == 0 or n == 1:
        return 1
    
    # 2. Recursive Step: n ko (n - 1) ke factorial se multiply karein
    return n * factorial(n - 1)

# Test the function:
print(factorial(5))  # Output: 120
print(factorial(0))  # Output: 1
```

```python
"""
4. Write a recursive function called sum_numbers that calculates 
the sum of numbers from 1 to n.
"""
def sum_numbers(n):
    # 1. Base Case: Agar n 1 hai, to sum sirf 1 hoga
    if n == 1:
        return 1
    
    # 2. Recursive Step: current number 'n' ko (n - 1) tak ke sum me jodna hai
    return n + sum_numbers(n - 1)

# Test the function:
print(sum_numbers(4))  # Output: 10
```

```python
""" 
5. Write a recursive function called power that takes two numbers base and
exponent and returns base raised to the power exponent.
"""
def power(base, exponent):
    # 1. Base Case: Kisi bhi number ki power 0 ho, to answer humesha 1 hota hai (Jaise 2^0 = 1)
    if exponent == 0:
        return 1
    
    # 2. Recursive Step: base ko (base ke [exponent - 1] power) se multiply karein
    return base * power(base, exponent - 1)

# Test the function:
print(power(2, 3))  # Output: 8 (kyunki 2 * 2 * 2 = 8)
print(power(5, 2))  # Output: 25 (kyunki 5 * 5 = 25)
```

```python
"""
9. Write a recursive function called count_digits that returns
the number of digit in a number.
"""
def count_digits(n):
    # Handle negative numbers
    n = abs(n)
    
    # Base case: if the number has only one digit
    if n < 10:
        return 1
    
    # Recursive case: 1 + count of digits in the remaining number
    return 1 + count_digits(n // 10)

# Test the function
print(count_digits(1465456454646))
```

```python
""" 
10. Write a recursive function called recursive_sum_list
that returns the sum of elements in a list.
"""
def recursive_sum_list(lst):
    # Base case: if the list is empty, return 0
    if len(lst) == 0:
        return 0
    
    # Recursive case: first element + sum of the rest of the list
    return lst[0] + recursive_sum_list(lst[1:])


# Test the function
numbers = [1, 2, 3, 4, 5]
print(recursive_sum_list(numbers))
```

```python
"""
11. Write a recursive function called 
reverse_number that reverses a number.
"""
def reverse_number(n):
    # Handle negative numbers
    sign = -1 if n < 0 else 1
    n = abs(n)
    
    # Base case: if single digit, return n
    if n < 10:
        return n * sign
    
    # Find the number of digits to calculate place value
    num_digits = len(str(n))
    last_digit = n % 10
    remaining = n // 10
    
    # Recursive case
    return sign * (last_digit * (10 ** (num_digits - 1)) + abs(reverse_number(remaining)))

# Test the function
print(reverse_number(12345))
```

```python
""" 
12. Write a recursive function called fibonacci that 
print the nth Fibonacci number.
"""
def fibonacci(n):
    # Base cases
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    
    # Recursive case
    return fibonacci(n - 1) + fibonacci(n - 2)

# Test the function
print(fibonacci(6))
```
