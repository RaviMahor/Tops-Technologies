```python
"""
Scenario 1: Converting a Set to a Tuple
Set elements are converted to a Tuple to allow indexing.
"""

my_set1 = {1, 2, 3, 4, 5}  # Original Set
my_list1 = tuple(my_set1)  # Converting to Tuple

print("Original Set:", my_set1)
print("Converted Tuple:", my_list1)
print("Type:", type(my_list1))
print("Element at index 1:", my_list1[1])  # Indexing works
```

```python
"""
Scenario 2: Converting a Tuple to a Set
Since sets do not support indexing, we access the element from the original tuple.
"""

my_set2 = (1, 2, 3, 4, 5)  # Original Tuple
my_list2 = set(my_set2)    # Converting Tuple to Set

print("Original Tuple:", my_set2)
print("Converted Set:", my_list2)
print("Type:", type(my_list2))
print("Element at index 1:", my_set2[1])  # Indexing original tuple se kiya hai
```

```python
"""
Scenario 3: Converting a Set to a List
Set elements are converted to a List to allow indexing.
"""

my_set = {1, 2, 3, 4, 5}  # Original Set
my_list = list(my_set)    # Converting Set to List

print("Original Set:", my_set)
print("Converted List:", my_list)
print("Type:", type(my_list))
print("Element at index 1:", my_list[1])  # Indexing works
```

```python
""" 
use if,elif and else to sumulate which type of income people can afford a car also write car names accordingly
1. If income is greater than 2000000, they can afford a luxury car like Ferrari
2. If income is greater than 1000000, they can afford a luxury car like Mercedes
3. If income is between 500000 and 1000000, they can afford a car like Honda but still needs to work hard to become wealthy
4. If income is less than 500000, they can afford an economy car like Maruti
"""

def format_income(amount):
    """Indian style: 15,00,000 — Easy to read"""
    s = str(amount)
    if len(s) <= 3:
        return f"₹{s}"
    result = s[-3:]
    s = s[:-3]
    while s:
        if len(s) <= 2:
            result = s + "," + result
            s = ""
        else:
            result = s[-2:] + "," + result
            s = s[:-2]
    return f"₹{result}"


income = int(input("Enter your income: "))
readable_income = format_income(income)

if income > 2000000:
    print(f"Your income is {readable_income} per year")
    print("You can afford a luxury car like Ferrari")
elif income > 1000000:
    print(f"Your income is {readable_income} per year")
    print("You can afford a luxury car like Mercedes")
elif income >= 500000:
    print(f"Your income is {readable_income} per year")
    print("You can afford a car like Honda but still need to work hard to become wealthy")
else:
    print(f"Your income is {readable_income} per year")
    print("You can afford an economy car like Maruti")
```
