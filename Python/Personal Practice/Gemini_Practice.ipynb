```python
""" 
Question 3 (Level: Hard)
Is baar hum aapke Session 7 (Conditionals) aur Session 8 (Loops) ko combine karenge.
Maan lijiye aapke paas kuch shehron ke temperatures (taapmaan) ka data ek list mein hai:
temps = [35, 42, 28, 50, 31, 45]

Aapko ek for loop aur if condition ka use karke ek code likhna hai jo yeh 3 step kare:
Ek nayi khali list banayein: high_temps = []

Ek for loop chalayein jo temps list ke har ek number ko check kare.

Agar temperature 40 ya usse zyada (>= 40) hai, toh us number ko high_temps wali nayi list mein .append() kar de.
Loop khatam hone ke baad, high_temps list ko print kare.
Expected Final Output:
[42, 50, 45]
"""
temps = [35, 42, 28, 50, 31, 45]
high_temps = []

# Step 1: Ek-ek karke temperature ko 'i' mein nikalen
for i in temps:
    # Step 2: Check karein kya 'i' 40 ya usse bada hai
    if i >= 40:
        # Step 3: Agar haan, toh use nayi list mein append kar dein
        high_temps.append(i)

print(high_temps)
```

```python
"""
💻 Advanced Challenge 2: The Billing Machine
Problem Statement:
Aapko ek function banana hai jiska naam hoga generate_bill. 
Yeh function bracket (parameter) mein ek list lega jisme shopping items ke prices honge.
Function ke andar aapko yeh logic likhna hai:
Us list ke saare prices ka Total (Sum) nikalna hai.
(Yaad karein, aapne loops aur accumulator total = 0 ka use karke yeh pehle kiya tha).
Total nikalne ke baad check karna hai: Agar total 1000 ya usse zyada (>= 1000) hai, toh print kare: Total bill is [total]. VIP Discount Applied!
Agar total 1000 se kam hai, toh print kare: Total bill is [total]. No discount."""
# --- Hissa 1: Function Banana ---
def generate_bill(user_cart):
    # Step 1: Loop se PEHLE ek khali variable banayein jisme hum paise jodenge
    total = 0
    
    # Step 2: List (user_cart) me se ek-ek price (num) bahar nikalein
    for num in user_cart:
        # Step 3: Purane 'total' mein naya 'num' jodein aur wapas 'total' mein save karein
        total = total + num 
        
    # Jab loop khatam ho jayega, toh 'total' mein sabka sum aa chuka hoga.
    # Ab hum check karenge ki total 1000 se bada hai ya nahi:
    if total >= 1000:
        print(f"Total bill is {total}. VIP Discount Applied!")
    else:
        print(f"Total bill is {total}. No discount.")


# --- Hissa 2: User se Input Lena (Yeh aapne ekdum sahi likha tha) ---
user_cart = []
item_price = -1

while item_price != 0:
    item_price = int(input("Enter item price (Enter 0 to stop): "))
    if item_price != 0:
        user_cart.append(item_price)

# --- Hissa 3: Function ko Call karna (Jo chhoot gaya tha) ---
# Humari user_cart list taiyar hai, ab hum isko apne function ke andar bhejenge
generate_bill(user_cart)
```

```python
""" 
🏧 The New Mega Challenge: "Smart ATM Machine" (Level: Hard)
Aapko ek bank ke liye Smart ATM ka code likhna hai jo cash deposit (paise jama) karta hai.
Is baar koi step-by-step hint nahi hai, sirf rules hain:
Aapke Code Ke Rules:
Ek while loop ka use karke user se lagatar notes ki value (jaise 500, 200, 100) input lijiye.
Jab user 0 daale, toh input lena band kar dijiye.
In sabhi daale gaye notes ko ek list mein save kijiye (jaise cash_list).
Ek naya function banaiye jiska naam ho: process_deposit(notes_list)

Is function ke andar ek for loop aur total = 0 ka use karke un sabhi notes ka Total Sum nikaliye.
Total nikalne ke baad function ke andar hi check kijiye:
Agar total 50,000 se zyada (> 50000) hai, toh print kijiye: Transaction Alert: Total is [total]. PAN Card is required!
Agar total 50,000 ya usse kam hai, toh print kijiye: Deposit Successful! Total is [total].
Sabse aakhir mein, apne function ko call kijiye aur usme user dwara banayi gayi list pass kar dijiye.
"""
# --- PART 1: User Input (Aapka 100% Sahi Code) ---
cash_deposite = []
cash_userinput = -1

while cash_userinput != 0:
    cash_userinput = int(input("Enter deposite amount (Enter 0 to stop): "))
    if cash_userinput != 0:
        cash_deposite.append(cash_userinput)

# --- PART 2: The Brain (Function) ---
def process_deposit(cash_deposite):
    total = 0  # 1. Gullak ab function ke ANDAR hai
    
    for i in cash_deposite:
        total = total + i  # 2. Poori list nahi, sirf current note (i) ko joda
        
    # 3. Dhyan dein! 'if' condition ab for loop ke BAHAR hai (pichhe khiska di gayi hai)
    # Taaki poora total nikalne ke baad hi check ho.
    if total > 50000:
        print(f"Transaction Alert: Total is {total}. PAN Card is required!")
    elif total <= 50000:
        print(f"Deposit Successful! Total is {total}.")

# --- PART 3: Call the function ---
# 4. Brackets mein apni list bhej kar function ko chalu karein
process_deposit(cash_deposite)
```
